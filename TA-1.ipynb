```python
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
import re
def clean_text(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'\W', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()
    cleaned_words = []
    for word in text.split():
        if word not in stopwords.words('english'):
            stemmed = ps.stem(word)
            cleaned_words.append(stemmed)
    return " ".join(cleaned_words)
```

```python
data['cleaned_review']=data['review_text'].apply(clean_text)
data
```

```python
print(data[['review_text', 'cleaned_review']].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(data['cleaned_review'])
Y = data['sentiment'].map({'Negative': 0, 'Neutral': 1, 'Positive': 2})
X, Y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,data['sentiment'],test_size=0.2,random_state=42)
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
review = input("Enter your experience/review: ")


cleaned_review = clean_text(review)


review_tfidf = vectorizer.transform([cleaned_review])

print("Predicted Sentiment:", model.predict(review_tfidf)[0])
```
